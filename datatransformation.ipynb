```python
import pandas as pd
import sqlite3

# Read CSV
df = pd.read_csv("Dataset_of_Diabetes.csv")

# Create SQLite database
conn = sqlite3.connect("Diabetes.db")

# Write data to table
df.to_sql("diabetes", conn, if_exists="replace", index=False)

conn.close()

print("Database created successfully: Diabetes.db")
```

Converting db into excel

```python
import sqlite3
import pandas as pd

# Connect to SQLite database
conn = sqlite3.connect("Diabetes.db")

# Read data from table
df = pd.read_sql("SELECT * FROM diabetes", conn)

# Export to Excel
df.to_excel("Diabetes.xlsx", index=False)

conn.close()

print("Excel file created successfully: Diabetes.xlsx")
```

Changing csv delimiters

```python
# csv to tsv
import pandas as pd

csv_file = "client.csv"
tsv_file = "client.tsv"

df = pd.read_csv(csv_file)
df.to_csv(tsv_file, sep='\t', index=False)

print(f"TSV file created successfully: {tsv_file}")
```
